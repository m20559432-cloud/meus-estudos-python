```python
identificacao = int(input("deseja realizar a operação com quantos números: "))
print(" 1 para soma, 2 para subtração, 3 para multiplicação e 4 para divisão")
operacao = int(input("digite o número da operação: "))

if identificacao == 2 and operacao == 1:
  a = float(input("digite o primeiro número: "))
  b = float(input("digite o segundo número: "))
  soma = a + b
  print(soma)
elif identificacao == 2 and operacao == 2:
  c = float(input("digite o primeiro número: "))
  d = float(input("digite o segundo número: "))
  print(c - d)
elif identificacao == 2 and operacao == 3:
  e = float(input("digite o primeiro número: "))
  f = float(input("digite o segundo número: "))
  print(e * f)
elif identificacao == 2 and operacao == 4:
  g = float(input("digite o primeiro número: "))
  h = float(input("digite o segundo número: "))
  print(g / h)
else:
  print("número não é válido")
```
