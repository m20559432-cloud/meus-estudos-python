```python
while True:
  n = int(input("digite um número par: "))
  if n % 2 == 0:
    print("você digitou um número par, obrigado!")
    break
  else:
    print("você digitou um número ímpar")
```

```python
n = [5, 8, 2, 7, 9, 1]
for n in range(1, 10):
  if n % 2 != 0:
    continue
  else:
    print(f"o número {n} é par")
```
